```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Solve in easy way: y[n] = x[n] + 0.8*y[n-1]

# 1) Create sample index and input signal
n = np.arange(0, 20)
x = np.ones(len(n), dtype=float)  # x[n] = 1

# 2) Prepare output and initial condition
y = np.zeros(len(n), dtype=float)
y[0] = 0.0

# 3) Apply difference equation sample by sample
alpha = 0.8
for i in range(1, len(n)):
    y[i] = x[i] + alpha * y[i - 1]

# 4) Show first values clearly
print(" n   x[n]    y[n]")
print("-------------------")
for i in range(10):
    print(f"{i:2d}   {x[i]:.1f}    {y[i]:.4f}")

# 5) Plot input and output
plt.figure(figsize=(7, 4))
plt.stem(n, x, linefmt='--', markerfmt='bo', basefmt=' ', label='Input x[n]')
plt.stem(n, y, linefmt='-', markerfmt='gs', basefmt=' ', label='Output y[n]')
plt.xlabel('n')
plt.ylabel('Amplitude')
plt.title('Easy Solution: y[n] = x[n] + 0.8y[n-1]')
plt.grid(True, alpha=0.4)
plt.legend()
plt.show()
```
