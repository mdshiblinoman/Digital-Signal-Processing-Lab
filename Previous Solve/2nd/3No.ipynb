```python
import numpy as np
import matplotlib.pyplot as plt

# Sampling setup
fs = 200
duration = 1.0
t = np.arange(0, duration, 1 / fs)

# Signals
a = np.sin(2 * np.pi * 10 * t)           # 10 Hz sine
b = np.sign(np.sin(2 * np.pi * 10 * t))  # 10 Hz square
c = np.sin(2 * np.pi * 20 * t)           # 20 Hz sine

# Correlations (full mode)
auto_a = np.correlate(a, a, mode='full')
cross_ab = np.correlate(a, b, mode='full')
cross_ac = np.correlate(a, c, mode='full')

# Lag axis
lags = np.arange(-len(a) + 1, len(a))

# Plot signals
plt.figure(figsize=(11, 7))
plt.subplot(3, 1, 1)
plt.plot(t, a, label='a: 10 Hz sine')
plt.grid(True)
plt.legend()

plt.subplot(3, 1, 2)
plt.plot(t, b, label='b: 10 Hz square', color='tab:orange')
plt.grid(True)
plt.legend()

plt.subplot(3, 1, 3)
plt.plot(t, c, label='c: 20 Hz sine', color='tab:green')
plt.xlabel('Time (s)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()

# Plot correlations
plt.figure(figsize=(11, 8))
plt.subplot(3, 1, 1)
plt.plot(lags, auto_a, label='Autocorrelation of a', color='tab:blue')
plt.grid(True)
plt.legend()

plt.subplot(3, 1, 2)
plt.plot(lags, cross_ab, label='Cross-correlation between a and b', color='tab:red')
plt.grid(True)
plt.legend()

plt.subplot(3, 1, 3)
plt.plot(lags, cross_ac, label='Cross-correlation between a and c', color='tab:purple')
plt.xlabel('Lag (samples)')
plt.grid(True)
plt.legend()

plt.tight_layout()
plt.show()
```
