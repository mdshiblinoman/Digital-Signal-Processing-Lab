```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Sampling setup
fs = 100          # sampling frequency (Hz)
N = 100           # number of samples (1 second)
t = np.arange(N) / fs

# Two sine waves
x_30 = np.sin(2 * np.pi * 30 * t)
x_70 = np.sin(2 * np.pi * 70 * t)

# DFT (one-sided magnitude spectrum)
freq = np.fft.rfftfreq(N, d=1/fs)
X30 = np.fft.rfft(x_30)
X70 = np.fft.rfft(x_70)
mag30 = np.abs(X30) / N
mag70 = np.abs(X70) / N

# Plot spectra
plt.figure(figsize=(10, 4))
plt.plot(freq, mag30, 'o-', label='30 Hz signal')
plt.plot(freq, mag70, 's--', label='70 Hz signal (aliases to 30 Hz)')
plt.title('Magnitude Spectrum Comparison (fs = 100 Hz)')
plt.xlabel('Frequency (Hz)')
plt.ylabel('Magnitude')
plt.xlim(0, fs/2)
plt.grid(True)
plt.legend()
plt.show()

# Print peak frequencies
peak30 = freq[np.argmax(mag30)]
peak70 = freq[np.argmax(mag70)]
print(f'Peak for 30 Hz sine: {peak30:.1f} Hz')
print(f'Peak for 70 Hz sine: {peak70:.1f} Hz')
```
